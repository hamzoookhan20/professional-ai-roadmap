````python
def create_cv_prompt(topic, audience, difficulty):

    template = """
You are a professional Computer Vision instructor.

Topic:
{topic}

Audience:
{audience}

Difficulty:
{difficulty}

Explain the topic using:

1. Simple definition
2. Why it matters
3. Intuition
4. Important terminology
5. Mathematics when relevant
6. Easy example
7. Intermediate example
8. Professional AI/CV example
9. Python/OpenCV implementation
10. Common mistakes
11. Best practices
12. Exercises

Make the explanation technically accurate.
Do not skip important concepts.
"""

    return template.format(
        topic=topic,
        audience=audience,
        difficulty=difficulty
    )
````

````python
def create_cv_prompt(topic, audience, difficulty, language="English"):

    template = """
You are a professional Computer Vision instructor.

Topic:
{topic}

Audience:
{audience}

Difficulty:
{difficulty}

Language:
{language}

Explain the topic using:

1. Simple definition
2. Why it matters
3. Intuition
4. Important terminology
5. Mathematics when relevant
6. Easy example
7. Intermediate example
8. Professional AI/CV example
9. Python/OpenCV implementation
10. Common mistakes
11. Best practices
12. Exercises

Make the explanation technically accurate.
Do not skip important concepts.
"""

    return template.format(
        topic=topic,
        audience=audience,
        difficulty=difficulty,
        language=language
    )
````
